```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test , y_train ,y_test = train_test_split( x,y, test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
print(y_pred[:5])
```

```python
comparison = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': y_pred
})
comparison.head()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
poly_comparison = pd.DataFrame({
    'Actual Price': y_test.values,
    'Predicted Price': poly_pred
})

poly_comparison.head()
```

```python
poly_mae = mean_absolute_error(y_test, poly_pred)
poly_mse = mean_squared_error(y_test, poly_pred)
poly_rmse = poly_mse ** 0.5
poly_r2 = r2_score(y_test, poly_pred)

print("Polynomial Regression")
print("MAE:", poly_mae)
print("MSE:", poly_mse)
print("RMSE:", poly_rmse)
print("R² Score:", poly_r2)
```

```python
results = pd.DataFrame({
    "Model": ["Multiple Linear Regression", "Polynomial Regression"],
    "MAE": [mae, poly_mae],
    "MSE": [mse, poly_mse],
    "RMSE": [rmse, poly_rmse],
    "R² Score": [r2, poly_r2]
})

results
```

```python
def predict_house_price(area, bedrooms, bathrooms, stories, parking):
    new_house = pd.DataFrame({
        'area': [area],
        'bedrooms': [bedrooms],
        'bathrooms': [bathrooms],
        'stories': [stories],
        'parking': [parking]
    })
    
    prediction = model.predict(new_house)
    
    return prediction[0]
```

```python
price = predict_house_price(2000, 3, 2, 2, 1)

print(f"Predicted House Price: ₹{price:,.2f}")
```

```python
price2 = predict_house_price(1500, 2, 2, 1, 1)

print(f"Predicted House Price: ₹{price2:,.2f}")
```
